```python
import pandas as pd

data = {
    "Name": ["Asha", "Rahul", "Priya", "Rahul", "Amit"],
    "Marks": [85, None, 92, None, 70],
    "City": ["Pune", "Mumbai", "Pune", "Mumbai", None]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

print("\nFirst Five Rows:")
print(df.head())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["City"] = df["City"].fillna("Unknown")

print("\nDuplicate Records:")
print(df.duplicated())

df = df.drop_duplicates()

result = df.loc[df["Marks"] >= 80]

print("\nCleaned Data:")
print(df)

print("\nStudents scoring 80 or more:")
print(result)

print("\nStatistical Summary:")
print(df.describe())
```
